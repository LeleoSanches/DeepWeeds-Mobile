```python
import pandas as pd
```

# Notebook para consolidação das tabelas
- Constroí a tabela de sumário e a tabela de resultados por classe

## Constroí dataframe de sumário dos modelos

```python
data_summary = pd.read_json("/home/leo/Documentos/mestrado/DeepWeeds-Mobile/models_summary.json", orient="index")
data_summary.drop(columns=["file", "path",'trainable','non_trainable','memory_MB', 'dtype_breakdown','input_h', 'input_w', 'FLOPs', 'bench_batch',
       'bench_runs', 'bench_warmup','lat_ms_p50', 'lat_ms_p95','fps_p50', 'fps_p95','inputs', 'outputs'], inplace=True)
data_summary = data_summary.reset_index(drop=True)
data_summary.rename(columns={
    "total": "Parâmetros (M)",
    "lat_ms_mean": "Latência Média (ms)",
    "fps_mean": "FPS Média",
    "model_name": "Modelo"
}, inplace=True)
data_summary["Parâmetros (M)"] = data_summary["Parâmetros (M)"] / 1_000_000
data_summary["Parâmetros (M)"] = data_summary["Parâmetros (M)"].round(2)
data_summary["Latência Média (ms)"] = data_summary["Latência Média (ms)"].round(2)
data_summary["GFLOPs"] = data_summary["GFLOPs"].round(2)
data_summary["GMAC"] = data_summary["GMAC"].round(2)
data_summary["FPS Média"] = data_summary["FPS Média"].round(2)


data_summary
```

## Constroí a tabela de resultados por classe

```python
df = pd.read_csv('results.csv')
df_classes = df.pivot_table(
    index="model",
    columns="class",
    values=["precision", "recall", "f1"]
)
df = df.rename(columns={"model": "Modelo", 
                        "class": "Classe", 
                        "precision": "Precisão", 
                        "recall": "Recall", 
                        "f1": "F1"})
df_classes.to_csv("results_byclass.csv", index=True)
df_classes
```

## Faz o merge da tabela resultados Macro

```python
df = pd.read_csv('results_macro.csv')
df["model"] = df["model"].str.replace("report_", "", regex=False)
df_macro = df.rename(columns={"model": "Modelo", 
                        "accuracy": "Acurácia Média", 
                        "macro_precision": "Precisão Média", 
                        "macro_recall": "Recall Médio", 
                        "macro_f1": "F1 Médio"})
df_macro
```

```python
df_final = df_macro.merge(
    data_summary,
    on="Modelo",
    how="inner"
)
df_final
```

## Faz o merge da tabela benchmark com os dataset real
- Note que há as métricas [REAL] que foram utilizados o dataset do deepweeds
- as outras métricas de FPS e latência utilizam Dummy

```python
data_bench = pd.read_csv("bench_out/bench_all.csv")
data_bench = data_bench.drop(columns=["file", "path", "input_h", "input_w", "batch" ,"steps",'lat_ms_p50', 'lat_ms_p95','fps_p50', 'fps_p95', 'preprocess_embutido'])
data_bench.rename(columns={
    "model_key": "Modelo",
    "lat_ms_mean": "[Real] Latência Média (ms)",
    "fps_mean": "[Real] FPS Média"
}, inplace=True)
data_bench["[Real] Latência Média (ms)"] = data_bench["[Real] Latência Média (ms)"].round(2)
```

```python
summary = data_bench.merge(
    df_final,
    on="Modelo",
    how="inner"

)
summary.to_csv("summary_results.csv", index=False)
```
